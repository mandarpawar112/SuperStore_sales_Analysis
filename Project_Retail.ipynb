# Stage 3: Exploratory Data Analysis (EDA)
Goal: Discover patterns, trends, and business problems using pandas and NumPy.

**A. Sales and profit analysis**

```python
# Calculate total sales, total profit, and total quantity sold.
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_quantity = df['Quantity'].sum()

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Total Quantity:", total_quantity)
```

```python
# Calculate the average sales amount per row.
df['Sales'].mean()
```

```python
# Calculate the average profit per row.
df['Profit'].mean()
```

```python
# Calculate the overall profit margin as total profit divided by total sales.
profit_margin = (df['Profit'].sum() / df['Sales'].sum()) * 100

print("Profit Margin:", round(profit_margin, 2), "%")
```

```python
#] Find the top 10 orders by total sales, aggregating rows by Order ID
top_10_orders = (
    df.groupby('Order ID')['Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_10_orders)
```

```python
# 6. Which Category has the highest total sales?
df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
```

```python
# 7. Which Category has the highest total profit?
df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
```

```python
# 8. Which Sub-Category has the highest total sales?
df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)
```

```python
# 9. Which Sub-Category has the lowest or most negative total profit?
df.groupby('Sub-Category')['Profit'].sum().sort_values()
```

```python
# 10. Identify the 10 most profitable products by total profit.
df.groupby('Product Name')['Profit'].sum().sort_values(ascending=False).head(10)
```

```python
# 11. Identify the 10 products with the highest total sales.
df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)
```

**C. Customer and regional analysis**

```python
# 12. Which Region generates the highest sales?
df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
```

```python
# 13. Which State generates the highest sales?
df.groupby('State')['Sales'].sum().sort_values(ascending=False)
```
